## 2. 문자열 길이 확인하기 📏

문자열의 길이를 알고 싶을 때는 `len()`을 사용합니다.

이때 `len()`은 **built-in function** 중 하나입니다.

```python
text = "Python is useful for English teachers."
print(text)
print(len(text))
```

### 설명 💡
- `len()`은 길이를 구하는 함수입니다.

## 3. 문자열을 소문자로 바꾸기 🔡

문자열에서는 대문자와 소문자가 섞여 있을 수 있습니다.  
이럴 때 `lower()`를 사용하면 모두 소문자로 바꿀 수 있습니다.

```python
text = "Python Is USEFUL For English Teachers."
print(text)
print(text.lower())
```

### 설명 💡
- `lower()`는 문자열을 모두 소문자로 바꿉니다.
- 나중에 단어 빈도 분석을 할 때 매우 자주 사용됩니다.

예를 들어 `Teacher`와 `teacher`를 같은 단어로 보고 싶을 때 유용합니다.

## 4. 문자열 나누기: split() ✂️

문장을 단어 단위로 나누고 싶을 때 `split()`을 사용할 수 있습니다.

```python
text = "Python is useful for English teachers."
words = text.split()

print(words)
print(type(words))
```

### 설명 💡
- `split()`은 문자열을 나누는 기능입니다.
- 보통 띄어쓰기 기준으로 나누어 줍니다.
- 결과는 문자열이 아니라 **list**가 됩니다.

즉, 문장 하나가 단어 목록으로 바뀌는 것입니다.

## 5. 문자열 바꾸기: replace() 🔄

문자열 안의 특정 부분을 다른 글자로 바꾸고 싶을 때는 `replace()`를 사용합니다.

```python
text = "I like English."
new_text = text.replace("English", "Python")

print(text)
print(new_text)
```

### 설명 💡
- `replace("English", "Python")`은
- `"English"`를 `"Python"`으로 바꾸라는 뜻입니다.

문자열 정리나 전처리에서 자주 쓰입니다.

## 7. Built-in function이란? 🛠️

**Built-in function** 은 파이썬에 원래 들어 있는 기본 함수입니다.

우리가 이미 본 built-in function 예:
- `print()`
- `len()`
- `type()`

즉, 따로 만들지 않아도 바로 사용할 수 있는 함수입니다.

```python
text = "hello"

print(text)
print(len(text))
print(type(text))
```

### 자주 보는 built-in function 예시 👀

- `print()` : 화면에 출력
- `len()` : 길이 구하기
- `type()` : 자료형 확인
- `str()` : 문자열로 바꾸기
- `int()` : 정수로 바꾸기
- `float()` : 실수로 바꾸기

```python
print(str(100))
print(int(3.9))
print(float(5))
```

### 설명 💡
- `str(100)` → 숫자 100을 문자열 `"100"`처럼 다룰 수 있게 바꿈
- `int(3.9)` → 3.9를 정수 3으로 바꿈
- `float(5)` → 5를 5.0으로 바꿈

이런 함수들은 파이썬이 원래 제공하는 기본 도구입니다.

## 8. Customized function이란? 🧩

**Customized function** 은 사용자가 직접 만드는 함수입니다.

즉,  
“내가 자주 하고 싶은 작업을 하나의 작은 도구로 만들어 두는 것”  
이라고 생각하면 됩니다.

```python
def say_hello():
    print("Hello!")
```

위 코드는 `say_hello()`라는 함수를 **만든 것**입니다.  
아직 실행한 것은 아닙니다.

```python
say_hello()
```

### 설명 💡
- `def`는 함수를 만들 때 사용합니다.
- `say_hello`는 함수 이름입니다.
- `say_hello()`라고 써야 실행됩니다.

즉, customized function은  
**내가 원하는 작업을 직접 만들어 놓은 함수**입니다.

## 9. 매개변수(parameter)가 있는 함수 📦

함수는 값을 받아서 작동할 수도 있습니다.

```python
def greet(name):
    print("Hello,", name)

greet("Jensen")
greet("students")
```

### 설명 💡
- `name`은 함수 안으로 들어오는 값입니다.
- 이처럼 함수는 입력값을 받아서 다른 결과를 만들 수 있습니다.

이 구조는 나중에 NLP 함수에서도 자주 보게 됩니다.

## 10. 영어교육 예시: 단어 수 세기 함수 📚

```python
def count_words(sentence):
    words = sentence.split()
    print("Word count:", len(words))

count_words("Python is useful for teachers.")
count_words("I like English class.")
```

### 설명 💡
이 함수는
1. 문장을 단어로 나누고
2. 단어 개수를 세고
3. 결과를 출력합니다

즉, 문자열 처리 + built-in function + customized function이 함께 쓰인 예입니다.

## 11. return이 있는 함수 ↩️

함수는 결과를 **돌려줄(return)** 수도 있습니다.

```python
def count_words(sentence):
    words = sentence.split()
    return len(words)

result = count_words("Python is useful for teachers.")
print(result)
```

### 설명 💡
- `print()`는 화면에 보여 주는 것
- `return`은 결과를 함수 밖으로 돌려주는 것

초보 단계에서는 이 차이를 완벽히 이해하지 않아도 괜찮습니다.

## 12. NLTK란? 📦

**NLTK**는 **Natural Language Toolkit**의 약자입니다.

파이썬에서 자연어처리(NLP)를 공부할 때 아주 오래전부터 널리 쓰인 대표적인 패키지입니다.

NLTK로 자주 하는 일:
- tokenization(문장/단어 나누기)
- stopwords 처리
- stemming
- part-of-speech tagging
- 간단한 corpus 다루기

## 13. NLTK에서 자주 보는 기능들 👀

대표적으로 아래 같은 기능을 자주 봅니다.

- `word_tokenize()` : 문장을 단어 단위로 나누기
- `sent_tokenize()` : 글을 문장 단위로 나누기
- `stopwords` : 불용어 목록 사용하기
- `FreqDist` : 단어 빈도 세기
- `pos_tag()` : 품사 태깅하기

오늘은 “이런 기능이 있구나”를 아는 것이 목표입니다.

## 14. NLTK 아주 가벼운 예시 🌱

Colab에서는 처음 사용할 때 리소스를 다운로드해야 하는 경우가 있습니다.  
아래 예시는 단어를 간단히 나누어 보는 예입니다.

```python
import nltk
nltk.download('punkt')

from nltk.tokenize import word_tokenize

text = "Python is useful for English teachers."
tokens = word_tokenize(text)

print(tokens)
```

### 설명 💡
- `nltk.download('punkt')`는 필요한 토큰화 자료를 받는 과정입니다.
- `word_tokenize()`는 문장을 더 세밀하게 나눌 수 있습니다.
- 우리가 앞에서 본 `split()`보다 NLP에 더 가까운 방식입니다.

## 15. NLTK에서 단어 빈도 보기 📊

```python
import nltk
nltk.download('punkt')

from nltk.tokenize import word_tokenize
from nltk import FreqDist

text = "English teachers use Python and teachers like Python."
tokens = word_tokenize(text.lower())

fdist = FreqDist(tokens)
print(fdist.most_common(5))
```

### 설명 💡
- `FreqDist`는 단어 빈도를 세는 도구입니다.
- 이런 기능은 나중에 vocabulary analysis나 text analysis에 연결될 수 있습니다.

## 16. spaCy란? ⚡

**spaCy**는 현대적인 NLP 작업에서 매우 널리 쓰이는 파이썬 패키지입니다.

NLTK보다 조금 더 **실용적이고 빠른 NLP 처리**에 많이 사용됩니다.

spaCy로 자주 하는 일:
- tokenization
- part-of-speech tagging
- lemmatization
- named entity recognition
- dependency parsing

## 17. spaCy의 주요 기능들 👀

spaCy에서 자주 보는 기능은 아래와 같습니다.

- tokenization : 단어 단위로 나누기
- POS tagging : 품사 정보 보기
- lemmatization : 기본형 찾기
- NER : 사람 이름, 장소, 기관명 같은 개체 찾기
- parsing : 문장 구조 분석하기

오늘은 세부 원리보다  
**“spaCy는 조금 더 강력한 NLP 패키지구나”**  
정도로 이해하면 충분합니다.

## 18. spaCy 아주 가벼운 예시 🌱

Colab에서는 spaCy와 언어 모델을 설치해야 할 수 있습니다.

```python
!pip -q install spacy
!python -m spacy download en_core_web_sm
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Python is useful for English teachers.")

for token in doc:
    print(token.text, token.pos_)
```

### 설명 💡
- `token.text`는 단어 자체
- `token.pos_`는 품사 정보입니다

즉, spaCy는 단어를 나누는 것뿐 아니라  
각 단어에 대한 문법 정보도 보여줄 수 있습니다.
