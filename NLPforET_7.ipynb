## 2. 필요한 자료 다운로드 📦

## 3. 두 텍스트 불러오기 📚

```python
from nltk.corpus import gutenberg

alice_raw = gutenberg.raw('carroll-alice.txt')
emma_raw = gutenberg.raw('austen-emma.txt')

print("Alice characters:", len(alice_raw))
print("Emma characters:", len(emma_raw))
```

## 4. 텍스트 일부 확인하기 👀

텍스트의 분위기가 어떻게 다른지 앞부분을 조금만 확인해 보겠습니다.

```python
print("=== Alice ===")
print(alice_raw[:500])

print("\n=== Emma ===")
print(emma_raw[:500])
```

## 5. 전처리 준비: 소문자 + 토큰화 ✂️

비교를 위해 두 텍스트를 비슷한 방식으로 처리하겠습니다.
- 소문자로 바꾸기
- 단어 단위로 나누기
- 알파벳 단어만 남기기

```python
from nltk.tokenize import word_tokenize

alice_tokens = word_tokenize(alice_raw.lower())
emma_tokens = word_tokenize(emma_raw.lower())

alice_words = [w for w in alice_tokens if w.isalpha()]
emma_words = [w for w in emma_tokens if w.isalpha()]

print("Alice words:", alice_words[:20])
print("Emma words:", emma_words[:20])
```

## 6. 전체 단어 수 비교 🔢

```python
print("Alice total words:", len(alice_words))
print("Emma total words:", len(emma_words))
```

## 7. 고유 단어 수 비교 🌟

고유 단어 수는 중복을 제거한 서로 다른 단어 종류의 수를 말합니다.

```python
alice_unique = set(alice_words)
emma_unique = set(emma_words)

print("Alice unique words:", len(alice_unique))
print("Emma unique words:", len(emma_unique))
```

## 8. Lexical diversity란? 🧠

**Lexical diversity**는 어휘 다양성을 간단히 보는 지표입니다.

아주 단순하게는 아래처럼 계산할 수 있습니다.

**고유 단어 수 / 전체 단어 수**

값이 높을수록 상대적으로 다양한 어휘를 사용했다고 볼 수 있습니다.

```python
alice_lex_div = len(alice_unique) / len(alice_words)
emma_lex_div = len(emma_unique) / len(emma_words)

print("Alice lexical diversity:", alice_lex_div)
print("Emma lexical diversity:", emma_lex_div)
```

### 설명 💡
이 값은 아주 간단한 지표입니다.  
완벽한 척도는 아니지만, 초보 단계에서는  
“어휘 다양성을 수치로 볼 수 있구나”  
를 이해하는 데 좋습니다.

## 9. 문장 단위로 나누기 ✂️

이번에는 평균 문장 길이를 보기 위해 문장 단위로 나누겠습니다.

```python
from nltk.tokenize import sent_tokenize

alice_sents = sent_tokenize(alice_raw)
emma_sents = sent_tokenize(emma_raw)

print("Alice number of sentences:", len(alice_sents))
print("Emma number of sentences:", len(emma_sents))
```

## 10. 평균 문장 길이 계산하기 📏

여기서는 아주 단순하게  
**문장당 평균 단어 수**를 계산해 보겠습니다.

```python
def average_sentence_length(sentences):
    sentence_lengths = []
    for sent in sentences:
        tokens = word_tokenize(sent)
        words = [w for w in tokens if w.isalpha()]
        sentence_lengths.append(len(words))
    return sum(sentence_lengths) / len(sentence_lengths)

alice_avg_sent_len = average_sentence_length(alice_sents)
emma_avg_sent_len = average_sentence_length(emma_sents)

print("Alice average sentence length:", alice_avg_sent_len)
print("Emma average sentence length:", emma_avg_sent_len)
```

### 설명 💡
이 값은 “한 문장에 평균적으로 몇 개의 단어가 있는가”를 보여 줍니다.

이를 통해 아주 거칠게나마
- 문장이 짧은 편인지
- 문장이 긴 편인지

를 비교할 수 있습니다.

## 11. 자주 나오는 단어 비교 📊

```python
from nltk import FreqDist

alice_fdist = FreqDist(alice_words)
emma_fdist = FreqDist(emma_words)

print("Alice top 15:")
print(alice_fdist.most_common(15))

print("\nEmma top 15:")
print(emma_fdist.most_common(15))
```

## 12. stopwords를 제거하고 다시 비교하기 🚫

기능어보다 내용어를 더 잘 보기 위해 stopwords를 제거해 보겠습니다.

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

alice_content = [w for w in alice_words if w not in stop_words]
emma_content = [w for w in emma_words if w not in stop_words]

alice_content_fdist = FreqDist(alice_content)
emma_content_fdist = FreqDist(emma_content)

print("Alice content words top 15:")
print(alice_content_fdist.most_common(15))

print("\nEmma content words top 15:")
print(emma_content_fdist.most_common(15))
```

### 설명 💡
이제 `the`, `and`, `of` 같은 단어가 줄어들고,  
각 텍스트의 핵심 내용어가 조금 더 잘 보일 수 있습니다.

## 13. 결과를 표처럼 정리해 보기 📋

```python
print("=== Comparison Summary ===")
print("Alice total words:", len(alice_words))
print("Emma total words:", len(emma_words))
print("Alice unique words:", len(alice_unique))
print("Emma unique words:", len(emma_unique))
print("Alice lexical diversity:", round(alice_lex_div, 4))
print("Emma lexical diversity:", round(emma_lex_div, 4))
print("Alice avg sentence length:", round(alice_avg_sent_len, 2))
print("Emma avg sentence length:", round(emma_avg_sent_len, 2))
```

## 14. 간단한 시각화: lexical diversity 비교 📈

```python
import matplotlib.pyplot as plt

labels = ["Alice", "Emma"]
values = [alice_lex_div, emma_lex_div]

plt.figure(figsize=(6, 4))
plt.bar(labels, values)
plt.ylabel("Lexical Diversity")
plt.title("Lexical Diversity Comparison")
plt.show()
```

## 16. spaCy 맛보기: POS 비율 아주 간단히 보기 ⚡

이제 spaCy를 이용해 아주 가볍게 품사 분포를 살펴보겠습니다.

여기서는 긴 전체 텍스트 대신,  
각 작품의 앞부분 일부만 분석합니다.  
초보 단계에서는 “spaCy로도 이런 것이 가능하구나”를 보는 정도면 충분합니다.

```python
!pip -q install spacy
!python -m spacy download en_core_web_sm
```

```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

alice_doc = nlp(alice_raw[:5000])
emma_doc = nlp(emma_raw[:5000])

alice_pos = Counter([token.pos_ for token in alice_doc if token.is_alpha])
emma_pos = Counter([token.pos_ for token in emma_doc if token.is_alpha])

print("Alice POS counts:", alice_pos)
print("Emma POS counts:", emma_pos)
```

### 설명 💡
- `token.pos_`는 spaCy가 붙인 품사 정보입니다.
- 여기서는 아주 대략적으로 어떤 품사가 많이 나오는지 살펴보는 예입니다.

예를 들어
- `NOUN`
- `VERB`
- `ADJ`
- `ADV`

같은 태그가 많이 보일 수 있습니다.

## 17. spaCy 결과를 조금 더 보기 👀

```python
print("Alice top POS:")
print(alice_pos.most_common())

print("\nEmma top POS:")
print(emma_pos.most_common())
```
