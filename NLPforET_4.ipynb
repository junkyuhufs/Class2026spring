## 1. Corpus(코퍼스)란? 🗂️

**Corpus**는 분석을 위해 모아 놓은 텍스트 자료 모음입니다.

예를 들어,
- 소설 모음
- 뉴스 기사 모음
- 학생 작문 모음
- 연설문 모음

등이 corpus가 될 수 있습니다.

NLP에서는 이런 corpus를 이용해
- 단어 빈도 보기
- 문장 길이 보기
- 자주 나오는 표현 찾기
- 문체 비교하기

같은 일을 할 수 있습니다.

## 2. Gutenberg corpus란? 📚

NLTK에는 여러 가지 예제 corpus가 들어 있습니다.  
그중 **Gutenberg corpus**는 고전 문학 텍스트 모음입니다.

예를 들어 아래 같은 텍스트가 포함되어 있습니다.

- Jane Austen 작품
- Shakespeare 작품
- Bible
- Alice in Wonderland

오늘은 그중 **`carroll-alice.txt`** 를 사용하겠습니다.

```python
import nltk
nltk.download('gutenberg')
```

## 3. Gutenberg corpus 안에 어떤 텍스트가 있는지 보기 👀

```python
from nltk.corpus import gutenberg

print(gutenberg.fileids())
```

### 설명 💡
`gutenberg.fileids()`는 Gutenberg corpus 안에 들어 있는 텍스트 파일 이름 목록을 보여 줍니다.

오늘은 그중에서 아래 파일을 사용할 것입니다.

- `carroll-alice.txt`

## 4. Alice 텍스트 불러오기 🐇

```python
alice_text = gutenberg.raw('carroll-alice.txt')

print(type(alice_text))
print(alice_text[:1000])
```

### 설명 💡
- `gutenberg.raw()`는 텍스트 전체를 **문자열(string)** 로 불러옵니다.
- `alice_text[:1000]`은 처음 1000글자만 보여주는 코드입니다.

긴 텍스트 전체를 한 번에 다 보는 대신,  
앞부분 일부만 확인하는 방법입니다.

## 5. 텍스트 길이 확인하기 📏

```python
print("Number of characters:", len(alice_text))
```

### 설명 💡
여기서 `len(alice_text)`는 **문자 수(character count)** 를 보여 줍니다.

즉, 이 텍스트가 얼마나 긴지 대략 확인할 수 있습니다.

## 6. 소문자로 바꾸기 🔡

텍스트 분석에서는 대문자와 소문자를 통일하는 경우가 많습니다.

예를 들어 `Alice`와 `alice`를 같은 단어로 보고 싶을 수 있습니다.

```python
alice_lower = alice_text.lower()

print(alice_lower[:500])
```

### 설명 💡
- `lower()`는 모든 글자를 소문자로 바꿉니다.
- 이것은 NLP 전처리(preprocessing)에서 매우 자주 하는 작업입니다.

## 7. 단어로 나누기: tokenization의 아주 기초 ✂️

문자열 전체를 분석하기보다,  
단어 단위로 나누어서 보는 경우가 많습니다.

가장 단순한 방법은 `split()`입니다.

```python
alice_words_simple = alice_lower.split()

print(type(alice_words_simple))
print(alice_words_simple[:30])
```

### 설명 💡
- `split()`은 띄어쓰기 기준으로 문자열을 나눕니다.
- 결과는 **list**입니다.
- 즉, 긴 문자열 하나가 단어 목록으로 바뀝니다.

이것은 아주 단순한 방법이지만, 초보자가 처음 이해하기에는 좋습니다.

## 8. 전체 단어 수 세기 🔢

```python
print("Total number of words:", len(alice_words_simple))
```

## 9. 고유 단어 수 세기 🌟

같은 단어가 여러 번 나오더라도,  
중복을 빼고 “서로 다른 단어 종류” 수를 보고 싶을 수 있습니다.

이럴 때 `set()`을 사용할 수 있습니다.

```python
unique_words = set(alice_words_simple)

print("Number of unique words:", len(unique_words))
```

### 설명 💡
- `set()`은 중복을 허용하지 않는 자료형입니다.
- 따라서 단어 목록을 `set()`으로 바꾸면 중복이 제거됩니다.

이렇게 하면
- 전체 단어 수(total words)
- 고유 단어 수(unique words)

를 비교할 수 있습니다.

## 10. 자주 나오는 단어 보기 📊

이제 단어 빈도를 간단히 살펴보겠습니다.

이를 위해 Python의 `Counter`를 사용합니다.

```python
from collections import Counter

word_counts = Counter(alice_words_simple)
print(word_counts.most_common(20))
```

### 설명 💡
- `Counter`는 항목이 몇 번 나왔는지 세어 주는 도구입니다.
- `most_common(20)`은 가장 자주 나오는 20개를 보여 줍니다.

이 결과를 보면  
관사, 대명사, 접속사 같은 기능어가 많이 나오는 것을 볼 수 있습니다.

## 11. 약간 더 정리된 전처리 해 보기 🧹

현재 결과에는 쉼표, 마침표 같은 문장부호가 붙어 있는 경우도 있습니다.

아주 간단하게 문장부호를 없애 보겠습니다.

```python
import re

clean_text = re.sub(r'[^a-z\s]', '', alice_lower)
clean_words = clean_text.split()

print(clean_words[:30])
```

### 설명 💡
- `re`는 정규표현식(regular expression) 관련 라이브러리입니다.
- 여기서는 영어 소문자(`a-z`)와 공백(`\s`)만 남기고 나머지를 제거했습니다.
- 이렇게 하면 문장부호와 숫자 등을 단순하게 제거할 수 있습니다.

초보 단계에서는  
“텍스트를 조금 더 깔끔하게 정리하는 과정이구나”  
정도로 이해하면 충분합니다.

## 12. 전처리 후 다시 단어 수 보기 🔍

```python
print("Total words after cleaning:", len(clean_words))
print("Unique words after cleaning:", len(set(clean_words)))
```

## 13. 전처리 후 자주 나오는 단어 보기 📈

```python
clean_word_counts = Counter(clean_words)
print(clean_word_counts.most_common(20))
```

### 설명 💡
전처리 전보다 조금 더 깔끔한 빈도표를 볼 수 있습니다.

다만 여전히 `the`, `and`, `to` 같은 아주 흔한 단어가 많이 나옵니다.  
이것은 다음 단계에서 **stopwords**와 연결될 수 있습니다.

## 14. 특정 단어가 몇 번 나오는지 보기 🔎

우리는 전체 빈도표만이 아니라,  
특정 단어가 몇 번 나오는지도 확인할 수 있습니다.

```python
print("alice:", clean_word_counts["alice"])
print("rabbit:", clean_word_counts["rabbit"])
print("queen:", clean_word_counts["queen"])
```

이런 식의 분석은 작품 속 핵심 인물이나 주제어를 확인하는 데 도움이 됩니다.
