## 2. pandas란? 🐼

**pandas**는 파이썬에서 표 형태 데이터를 다루는 대표적인 라이브러리입니다.

pandas를 사용하면 아래 같은 일을 쉽게 할 수 있습니다.

- CSV 파일 읽기
- 표 형태로 데이터 보기
- 특정 열(column) 선택하기
- 평균, 합계, 개수 구하기
- 조건에 맞는 행(row)만 보기
- 간단한 정렬과 필터링 하기

초보 단계에서는  
**“엑셀 같은 표를 파이썬으로 다루는 도구”**  
라고 이해하면 좋습니다.

````python
import pandas as pd
````

## 3. 먼저 아주 작은 예제 데이터 만들기 ✨

실제 CSV 파일을 올리기 전에,  
수업에서는 먼저 작은 표를 직접 만들어 보면 이해가 쉽습니다.

````python
data = {
    "name": ["Alice", "Ben", "Chris", "Dina"],
    "score": [90, 85, 95, 88],
    "word_count": [120, 98, 145, 110],
    "topic": ["travel", "school", "music", "books"]
}

df = pd.DataFrame(data)
df
````

### 설명 💡
- `pd.DataFrame(data)`는 표 형태 데이터를 만듭니다.
- `df`는 DataFrame의 줄임말로 많이 사용합니다.
- DataFrame은 pandas에서 가장 중요한 자료형입니다.

즉, 이제 우리는 작은 표를 파이썬 안에 만든 것입니다.

## 4. DataFrame의 기본 정보 보기 👀

````python
print(type(df))
print(df.shape)
print(df.columns)
````

### 설명 💡
- `type(df)` → 자료형 확인
- `df.shape` → (행 개수, 열 개수)
- `df.columns` → 열 이름 보기

이렇게 표의 기본 구조를 먼저 확인할 수 있습니다.

## 5. 특정 열(column) 보기 📌

표 전체가 아니라 특정 열만 보고 싶을 때가 많습니다.

````python
print(df["name"])
print()
print(df["score"])
````

### 설명 💡
- `df["name"]` → 이름 열 보기
- `df["score"]` → 점수 열 보기

즉, 열 이름을 사용해 필요한 정보만 꺼낼 수 있습니다.

## 6. 평균 구하기 📊

점수나 단어 수처럼 숫자 데이터가 있으면 평균을 구할 수 있습니다.

````python
print("Average score:", df["score"].mean())
print("Average word count:", df["word_count"].mean())
````

### 설명 💡
- `mean()`은 평균을 구하는 함수입니다.
- pandas에서는 숫자 열에 대해 이런 계산을 쉽게 할 수 있습니다.

영어교육에서는
- 평균 점수
- 평균 문장 길이
- 평균 단어 수

같은 값을 자주 보게 됩니다.

## 7. 가장 큰 값, 가장 작은 값 보기 🔍

````python
print("Highest score:", df["score"].max())
print("Lowest score:", df["score"].min())
````

## 8. 조건에 맞는 행만 보기 🚦

예를 들어 점수가 90 이상인 학생만 보고 싶을 수 있습니다.

````python
high_score_df = df[df["score"] >= 90]
high_score_df
````

### 설명 💡
이 코드는  
`score`가 90 이상인 행만 남깁니다.

즉, **조건 filtering**의 기초입니다.

## 9. 정렬하기 ↕️

점수 순으로 정렬해 보겠습니다.

````python
df_sorted = df.sort_values("score", ascending=False)
df_sorted
````

### 설명 💡
- `sort_values("score")` → score 기준 정렬
- `ascending=False` → 큰 값부터 내림차순 정렬

이렇게 하면 상위 점수 학생을 바로 볼 수 있습니다.

## 10. 기술통계 한 번에 보기 📋

숫자 열이 여러 개 있을 때 `describe()`를 쓰면 편리합니다.

````python
df.describe()
````

### 설명 💡
`describe()`는 아래 같은 정보를 한 번에 보여 줍니다.

- 개수(count)
- 평균(mean)
- 표준편차(std)
- 최소값(min)
- 최대값(max)

초보 단계에서는  
“숫자 데이터 요약표를 보여 주는 기능”  
으로 이해하면 충분합니다.

## 11. 텍스트 열도 다룰 수 있다 ✍️

pandas는 점수 같은 숫자뿐 아니라,  
텍스트 열도 함께 다룰 수 있습니다.

````python
data2 = {
    "student": ["A", "B", "C"],
    "writing": [
        "I like English class because it is fun.",
        "English is important for my future job.",
        "I want to improve my speaking and writing."
    ]
}

df_text = pd.DataFrame(data2)
df_text
````

## 12. 텍스트 길이 세기 📏

학생 작문 데이터가 있으면,  
각 글의 길이를 간단히 셀 수도 있습니다.

````python
df_text["char_count"] = df_text["writing"].apply(len)
df_text
````

### 설명 💡
- `apply(len)`은 각 행의 글자 수를 계산합니다.
- 새로운 열 `char_count`를 만들어 저장했습니다.

즉, 기존 표에 새로운 분석 결과를 추가할 수 있습니다.

## 13. 단어 수 세기 🔢

````python
df_text["word_count"] = df_text["writing"].apply(lambda x: len(x.split()))
df_text
````

### 설명 💡
- `x.split()`으로 문장을 단어 목록으로 나누고
- `len()`으로 단어 수를 셉니다.

이것은 학생 writing data를 아주 기초적으로 분석하는 좋은 출발점입니다.

## 14. 가장 긴 writing 찾기 🔎

````python
df_text.sort_values("word_count", ascending=False)
````
