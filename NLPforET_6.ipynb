## 2. 필요한 자료 다운로드 📦

이번 시간에는 Gutenberg corpus와 tokenization, POS tagging 관련 자료를 사용합니다.

```python
import nltk

nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

## 3. Alice 텍스트 다시 불러오기 🐇

```python
from nltk.corpus import gutenberg
from nltk.tokenize import word_tokenize

alice_text = gutenberg.raw('carroll-alice.txt').lower()
tokens = word_tokenize(alice_text)

print(tokens[:30])
```

## 4. POS tagging이란? 🏷️

**POS tagging**은 각 단어에 **품사(part of speech)** 정보를 붙이는 작업입니다.

예를 들어,
- noun (명사)
- verb (동사)
- adjective (형용사)
- adverb (부사)

같은 정보를 단어에 붙여 주는 것입니다.

즉, 단어가 문장 안에서 어떤 역할을 하는지 보는 방법입니다.

## 5. 단어에 품사 태그 붙여 보기 ✍️

```python
sample_tokens = tokens[:20]
tagged_sample = nltk.pos_tag(sample_tokens)

print(tagged_sample)
```

### 설명 💡
출력 결과는 보통 이런 형태입니다.

- `('alice', 'NN')`
- `('was', 'VBD')`
- `('very', 'RB')`

여기서
- 첫 번째는 단어
- 두 번째는 품사 태그

입니다.

## 6. 자주 보는 품사 태그 몇 가지 👀

초보 단계에서는 아래 정도만 알아도 충분합니다.

- `NN` : noun, 명사
- `NNS` : plural noun, 복수명사
- `VB` : verb, 동사 원형
- `VBD` : past tense verb, 과거형 동사
- `JJ` : adjective, 형용사
- `RB` : adverb, 부사
- `PRP` : personal pronoun, 인칭대명사

오늘은 태그를 모두 외우는 것이 아니라  
**“아, 단어마다 품사 정보를 붙일 수 있구나”**  
를 이해하는 것이 중요합니다.

## 7. 품사 태깅 결과를 조금 더 보기 🔍

```python
for item in tagged_sample:
    print(item)
```

## 8. 특정 품사만 골라 보기: 명사와 형용사 🧺

품사 태깅 결과에서 특정 품사만 골라 볼 수도 있습니다.

```python
tagged_tokens = nltk.pos_tag(tokens[:200])

nouns = [word for word, tag in tagged_tokens if tag.startswith('NN')]
adjectives = [word for word, tag in tagged_tokens if tag.startswith('JJ')]

print("Nouns:", nouns[:20])
print("Adjectives:", adjectives[:20])
```

### 설명 💡
- `tag.startswith('NN')` → 명사 계열 찾기
- `tag.startswith('JJ')` → 형용사 계열 찾기

이런 방식으로 텍스트 안에서  
어떤 종류의 단어가 많이 쓰이는지 살펴볼 수 있습니다.

## 9. concordance란? 📚

**Concordance**는 특정 단어가  
텍스트 안에서 **어떤 맥락(context)** 으로 쓰이는지 보여 주는 기능입니다.

즉, 단어 하나를 중심으로  
앞뒤에 어떤 말이 오는지를 여러 줄로 보여 줍니다.

이것은 코퍼스 언어학에서도 매우 중요한 기초 도구입니다.

참고. [COCA (Corpus of Comtemporary English corpus](https://www.english-corpora.org/coca/)


```python
from nltk.text import Text

alice_nltk_text = Text(tokens)
```

## 10. concordance 해 보기: alice 🔎

```python
alice_nltk_text.concordance("alice", lines=10)
```

### 설명 💡
이 결과를 보면 `alice`라는 단어가  
문장 속에서 어떤 맥락으로 쓰이는지 볼 수 있습니다.

즉,
- 어떤 동사와 함께 오는지
- 어떤 표현 근처에 자주 나오는지
- 어떤 장면에서 많이 등장하는지

를 살펴볼 수 있습니다.

## 11. 다른 단어도 concordance 해 보기 👀

```python
alice_nltk_text.concordance("rabbit", lines=10)
```

## 12. collocation이란? 🤝

**Collocation**은 함께 자주 나타나는 단어 조합을 말합니다.

예를 들어 영어에서는
- strong tea
- make a decision
- high school

같이 특정 단어들이 함께 자주 나타나는 경우가 있습니다.

이런 단어 결합을 보는 것은  
어휘 학습과 영어교육에서도 매우 중요합니다.

## 13. Alice 텍스트의 collocation 보기 🌟

```python
alice_nltk_text.collocations()
```

### 설명 💡
출력 결과에는 함께 자주 나타나는 표현들이 보입니다.

이것은 단순히 단어를 하나씩 보는 것이 아니라,  
**단어들이 함께 어떻게 쓰이는지** 보는 방법입니다.

## 14. similar()로 비슷한 맥락의 단어 보기 🔄

NLTK에서는 어떤 단어와 **비슷한 맥락에서 쓰이는 단어**를 볼 수도 있습니다.

```python
alice_nltk_text.similar("alice")
```

### 설명 💡
이 기능은 완전히 같은 뜻의 단어를 찾는 것은 아닙니다.  
대신 **비슷한 문맥 자리**에 오는 단어를 보여 줍니다.

즉, “이 단어와 비슷한 위치에서 쓰이는 단어가 무엇인가?”를 보는 기능입니다.

## 15. common_contexts()로 공통 맥락 보기 🧠

```python
alice_nltk_text.common_contexts(["alice", "she"])
```

이 기능은 두 단어가 어떤 비슷한 문맥에서 쓰이는지를 보여 줍니다.  
초보 단계에서는 결과가 조금 낯설 수 있지만,  
**맥락 비교 도구**라고 이해하면 충분합니다.

## 16. 영어교육 관점에서 왜 중요할까? 🎓

이런 분석은 영어교육에서 아래처럼 활용될 수 있습니다.

- 특정 단어의 실제 사용 맥락 보여 주기
- collocation 학습 자료 만들기
- reading text의 핵심 표현 찾기
- 문맥 속 어휘 학습 활동 설계하기
- learner corpus에서 부자연스러운 결합 찾기

즉, 단어를 “뜻”만이 아니라  
**맥락과 결합** 속에서 가르치는 데 도움이 됩니다.
