## 1. 변수(variable)란? 📦

변수(variable)는 **값을 저장해 두는 이름표**입니다.

예를 들어,
- 이름
- 나이
- 점수
- 문장
- 단어 목록

같은 값을 저장해 두었다가 나중에 다시 꺼내 쓸 수 있습니다.

즉, 변수는 **값을 담아 두는 상자**처럼 생각하면 됩니다.

```python
name = "Jensen"
print(name)
```

### 설명 💡
- `name`은 변수 이름입니다.
- `"Jensen"`라는 값을 `name`에 저장했습니다.
- `print(name)`은 저장된 값을 화면에 보여줍니다.

즉, 변수는 값을 한 번 저장해 두고 여러 번 다시 쓸 수 있게 해 줍니다.

## 2. 변수에 숫자도 저장할 수 있다 🔢

```python
age = 25
print(age)
```

변수에는 글자뿐 아니라 숫자, 참/거짓, 목록 등 여러 종류의 값을 저장할 수 있습니다.

## 3. Data type(자료형)이란? 🧩

자료형(data type)은 **데이터의 종류**를 말합니다.

예를 들어,
- 글자인지
- 숫자인지
- 소수인지
- 참/거짓인지
- 여러 값을 모아 둔 목록인지
- 이름표와 값으로 저장된 자료인지

를 구분하는 것입니다.

파이썬에서는 데이터의 종류에 따라 처리 방식이 달라집니다.

## 4. 대표적인 자료형 overview 👀

오늘 가장 먼저 볼 자료형은 아래 6가지입니다.

- `str` : 문자열(string), 글자
- `int` : 정수(integer)
- `float` : 실수(float), 소수점이 있는 숫자
- `bool` : 불리언(boolean), 참/거짓
- `list` : 여러 값을 순서대로 저장하는 목록
- `dict` : key와 value로 저장하는 자료형

```python
text = "hello"
age = 25
score = 3.14
is_ready = True
fruits = ["apple", "banana", "grape"]
student = {"name": "Jensen", "major": "English Education"}

print(text)
print(age)
print(score)
print(is_ready)
print(fruits)
print(student)
```

### 설명 💡
- `"hello"` 는 **string**
- `25` 는 **int**
- `3.14` 는 **float**
- `True` 는 **bool**
- `["apple", "banana", "grape"]` 는 **list**
- `{"name": "Jensen", "major": "English Education"}` 는 **dict**

즉, 변수 안에는 다양한 자료형의 값이 들어갈 수 있습니다.

## 5. type()으로 자료형 확인하기 🔍

파이썬에서는 `type()` 함수를 사용해 자료형을 확인할 수 있습니다.

```python
print(type("hello"))
print(type(25))
print(type(3.14))
print(type(True))
print(type(["apple", "banana"]))
print(type({"name": "Jensen"}))
```

### 설명 💡
`type()`은  
“이 값은 어떤 종류의 데이터인가?”  
를 보여주는 함수입니다.

## 6. string(문자열) ✍️

문자열은 글자 데이터입니다.  
단어, 문장, 문단 등을 저장할 때 사용합니다.

```python
sentence = "Python is useful for English teachers."
print(sentence)
print(type(sentence))
```

영어교육과 NLP에서는 문자열을 아주 많이 다룹니다.  
학생 문장, reading passage, vocabulary items 등이 모두 문자열입니다.

## 7. int(정수)와 float(실수) 🔢

숫자에는 대표적으로 두 가지가 있습니다.

- `int` : 정수
- `float` : 소수점이 있는 수

```python
a = 10
b = 3.5

print(a)
print(type(a))

print(b)
print(type(b))
```

### 설명 💡
- `10`은 정수이므로 `int`
- `3.5`는 소수이므로 `float`

```python
print(10 + 5)
print(3.5 + 1.2)
```

## 8. bool(불리언) ✅❌

불리언은 **참(True) / 거짓(False)** 를 나타내는 자료형입니다.

조건문에서 매우 중요합니다.


```python
is_student = True
is_teacher = False

print(is_student)
print(is_teacher)
print(type(is_student))
```

## 9. dict(딕셔너리) 🗂️

딕셔너리는 **이름표(key)와 값(value)** 을 짝지어 저장하는 자료형입니다.

```python
student = {
    "name": "Jensen",
    "major": "English Education",
    "year": 2026
}

print(student)
print(type(student))
```

딕셔너리는 이런 질문에 잘 맞습니다.

- 이름은?
- 전공은?
- 학년은?

즉, **항목명 + 값** 형태로 저장하고 싶을 때 유용합니다.

## 10. list(리스트) 📝

리스트는 **여러 값을 한 번에 저장하는 자료형**입니다.

```python
colors = ["red", "blue", "green"]
print(colors)
print(type(colors))
```

리스트는 대괄호 `[]`를 사용합니다.  
값 여러 개를 쉼표로 구분해서 넣습니다.

영어교육에서는
- 단어 목록
- 학생 이름 목록
- 문장 목록
- 응답 목록

등을 저장할 때 자주 씁니다.

## 11. 리스트 안의 값 하나 보기 👀

```python
words = ["teacher", "student", "classroom", "language"]

print(words[0])
print(words[1])
print(words[2])
```

파이썬은 첫 번째 항목을 `0`부터 셉니다.

- `words[0]` → 첫 번째 값
- `words[1]` → 두 번째 값
- `words[2]` → 세 번째 값

## 12. 리스트 길이 구하기 📏

```python
words = ["teacher", "student", "classroom", "language"]
print(len(words))
```

`len(words)`는 리스트 안에 들어 있는 항목 수를 보여 줍니다.

## 13. 리스트에 값 추가하기 ➕

```python
words = ["teacher", "student", "classroom"]
words.append("English")
print(words)
```

## 14. 반복문(for)이란? 🔁

반복문은 **같은 작업을 여러 번 자동으로 실행하는 문법**입니다.

리스트에 들어 있는 값을 하나씩 처리할 때 자주 사용합니다.

```python
words = ["teacher", "student", "classroom"]

for word in words:
    print(word)
```

### 설명 💡
이 코드는
- 리스트에서 값을 하나씩 꺼내고
- 그 값을 `word`에 넣고
- 출력합니다

즉, 같은 동작을 자동으로 반복합니다.

## 15. 들여쓰기(indentation) ↘️

파이썬에서는 반복문 뒤의 코드를 **들여쓰기** 해야 합니다.

보통 4칸 들여쓰기합니다.

```python
for i in [1, 2, 3]:
    print(i)
```

들여쓰기가 맞지 않으면 오류가 날 수 있습니다.  
파이썬은 들여쓰기를 매우 중요하게 생각합니다.

## 16. range()로 반복하기 🔢

```python
for i in range(5):
    print(i)
```

`range(5)`는 보통 `0, 1, 2, 3, 4`를 차례로 만든다고 생각하면 됩니다.

## 17. 영어교육 예시: 단어 목록 출력하기 📚

```python
vocab_list = ["read", "write", "speak", "listen"]

for word in vocab_list:
    print("Word:", word)
```

## 18. 조건문(if)이란? 🚦

조건문은 **상황에 따라 다른 코드를 실행하는 문법**입니다.

즉,  
“만약 이것이 맞으면 이렇게 하고, 아니면 저렇게 하라”  
라는 구조입니다.

```python
score = 85

if score > 80:
    print("Good job!")
```

이 코드는  
“만약 점수가 80보다 크면 Good job!을 출력하라”  
는 뜻입니다.

## 19. if - else ↔️

```python
score = 70

if score > 80:
    print("Excellent")
else:
    print("Keep practicing")
```

조건이 맞으면 `if` 아래 코드가 실행되고,  
맞지 않으면 `else` 아래 코드가 실행됩니다.

## 20. 영어교육 예시: 문장 길이 판단하기 ✏️

```python
sentence = "I like English."
length = len(sentence)

print("Length:", length)

if length > 15:
    print("This is a long sentence.")
else:
    print("This is a short sentence.")
```
