## 오늘 수업 목표

오늘은 아래 5가지를 이해하면 충분합니다.

1. **변수(variable)** 가 무엇인지 안다.  
2. **문자열(string)** 과 **숫자(number)** 의 차이를 안다.

## 1. 변수(variable)란?

변수는 **값을 저장해 두는 이름표**라고 생각하면 됩니다.

예를 들어 아래 코드를 봅시다.

```python
name = "Tom"
print(name)
```

### 설명
- `name`은 변수 이름입니다.
- `"Junkyu"`라는 값을 `name`에 저장했습니다.
- `print(name)`은 저장된 값을 화면에 보여줍니다.

즉, 변수는  
**값을 담아 두었다가 나중에 다시 쓰기 위한 상자**처럼 이해하면 됩니다.

## 2. 문자열(string) 저장하기

문자열은 글자 데이터입니다.  
문자열은 보통 따옴표로 묶습니다.

```python
message = "Hello, Python!"
print(message)
```

### 설명
- `"Hello, Python!"`은 문자열입니다.
- 문자열은 영어 문장, 단어, 한국어 문장 등 **글자로 된 정보**입니다.

## 3. 숫자(number) 저장하기

이번에는 숫자를 저장해 봅시다.

```python
age = 25
print(age)
```

### 설명
- `25`는 숫자입니다.
- 숫자는 계산에 사용할 수 있습니다.
- 문자열과 숫자는 겉보기에는 단순해 보여도 파이썬에서는 다르게 처리됩니다.

### 설명
- `25`는 숫자입니다.
- 숫자는 계산에 사용할 수 있습니다.
- 문자열과 숫자는 겉보기에는 단순해 보여도 파이썬에서는 다르게 처리됩니다.

```python
print("3")
print(3)
```

### 설명
- `"3"`은 글자입니다.
- `3`은 숫자입니다.

즉,
- 문자열은 **읽는 대상**
- 숫자는 **계산할 수 있는 대상**

이라고 생각하면 이해하기 쉽습니다.

## 5. 변수에 저장한 값을 다시 사용하기

변수는 한 번 저장하고 끝나는 것이 아니라,  
나중에 다시 꺼내서 사용할 수 있습니다.

```python
teacher = "English teacher"
print("I am an", teacher)
print(teacher)
```

### 설명
- `teacher`라는 변수에 `"English teacher"`를 저장했습니다.
- 그 값을 여러 번 다시 쓸 수 있습니다.
- 이것이 변수의 가장 큰 장점입니다.

## 6. 숫자로 계산하기

숫자를 변수에 저장하면 계산도 할 수 있습니다.

```python
a = 10
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

### 설명
- `+` : 더하기
- `-` : 빼기
- `*` : 곱하기
- `/` : 나누기

파이썬은 계산기처럼 사용할 수도 있습니다.

## 7. 문자열 연결하기

문자열끼리는 이어 붙일 수도 있습니다.

first_name = "Harry"
last_name = "Porter"

full_name = first_name + " " + last_name
print(full_name)

### 설명
문자열에서는 `+`가 숫자 계산이 아니라  
**글자 이어 붙이기** 역할을 합니다.

즉 `+`는 데이터 종류에 따라 다르게 보일 수 있습니다.

## 8. 영어교육 예시 1: 학생 문장 저장하기

영어교육에서는 문자열을 많이 다루게 됩니다.  
예를 들어 학생 문장을 변수에 저장할 수 있습니다.

```python
student_sentence = "I like studying English."
print(student_sentence)
```

이렇게 저장한 문장은 나중에
- 길이 세기
- 단어 나누기
- 자주 쓰는 표현 찾기
같은 분석으로 이어질 수 있습니다.

## 9. 길이(length) 확인하기

문자열의 길이를 확인할 때는 `len()`을 사용할 수 있습니다.

```python
student_sentence = "I like studying English."
print(student_sentence)
print(len(student_sentence))
```

### 설명
- `len()`은 길이를 구합니다.
- 지금은 우선 **문자 수를 보여주는 기능**으로 이해하면 됩니다.

## 10. 변수 이름 바꾸기와 다시 저장하기

변수에 새로운 값을 다시 넣을 수도 있습니다.

```python
topic = "reading"
print(topic)

topic = "writing"
print(topic)
```

### 설명
- 처음에는 `reading`이 저장됩니다.
- 나중에 `writing`으로 바꾸면, 이전 값 대신 새 값이 저장됩니다.

즉, 변수는 **상황에 따라 업데이트되는 저장 공간**입니다.

## 11. 간단한 함수(function) 맛보기

함수는 **어떤 작업을 해 주는 작은 도구**라고 생각하면 됩니다.

지금까지 사용한 `print()`와 `len()`도 함수입니다.

```python
text = "Python is useful."
print(text)
print(len(text))
```

### 설명
- `print()`는 보여주는 함수
- `len()`은 길이를 구하는 함수

즉, 함수는 **입력값을 받아서 어떤 일을 수행하는 장치**입니다.

## 12. 아주 간단한 사용자 함수 만들기

이번에는 우리가 직접 아주 쉬운 함수를 하나 만들어 보겠습니다.

```python
def greet():
    print("Hello!")
```

위 코드는 `greet()`라는 이름의 함수를 만든 것입니다.  
하지만 **만들기만 하고 아직 실행한 것은 아닙니다.**

```python
greet()
```

### 설명
- `def`는 함수를 만들 때 사용하는 표현입니다.
- `greet`는 함수 이름입니다.
- `greet()`라고 써야 실제로 실행됩니다.

오늘은 함수의 문법을 완전히 익히는 것이 목표가 아니라,  
**“아, 이런 것도 함수구나”** 정도를 느끼면 충분합니다.

## 13. 영어교육 예시 2: 간단한 응원 메시지 함수

아주 간단한 함수는 이런 식으로 만들 수 있습니다.

```python
def cheer():
    print("You can do it!")
    print("Keep studying English!")

cheer()
```

이처럼 함수는  
같은 작업을 여러 번 반복할 때 편리합니다.
