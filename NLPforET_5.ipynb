## 2. 필요한 NLTK 자료 다운로드 📦

이번 시간에는 Gutenberg corpus와 stopwords, punkt tokenizer를 사용합니다.

```python
import nltk

nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('stopwords')
```

## 3. Alice 텍스트 다시 불러오기 🐇

```python
from nltk.corpus import gutenberg

alice_text = gutenberg.raw('carroll-alice.txt')
alice_text = alice_text.lower()

print(alice_text[:500])
```

## 4. split()으로 나누기 vs word_tokenize()로 나누기 ✂️

지난 시간에는 `split()`을 사용했습니다.  
이번에는 NLTK의 `word_tokenize()`도 사용해 보겠습니다.

두 방법은 비슷해 보이지만 결과가 조금 다를 수 있습니다.

```python
simple_words = alice_text.split()

from nltk.tokenize import word_tokenize
tokenized_words = word_tokenize(alice_text)

print("split() result:")
print(simple_words[:30])

print("\nword_tokenize() result:")
print(tokenized_words[:30])
```

### 설명 💡
- `split()`은 띄어쓰기 기준으로 아주 단순하게 나눕니다.
- `word_tokenize()`는 NLP에서 좀 더 자주 쓰이는 방식으로 텍스트를 나눕니다.
- 문장부호 처리 등에서 차이가 나타날 수 있습니다.

즉, `word_tokenize()`가 조금 더 **자연어처리에 가까운 tokenization**이라고 볼 수 있습니다.

## 5. tokenization이란? 🧩

**Tokenization**은 텍스트를 작은 단위로 나누는 과정입니다.

예를 들어,
- 문장 단위로 나누기
- 단어 단위로 나누기

같은 작업이 tokenization입니다.

NLP에서는 거의 모든 분석의 출발점이 됩니다.

## 6. stopwords란? 🚫

**Stopwords**는 너무 자주 나와서  
분석에서 큰 의미를 주지 않는 경우가 많은 단어들입니다.

예를 들어 영어에서는 아래 같은 단어가 자주 stopwords로 간주됩니다.

- the
- and
- is
- to
- of

이런 단어들은 거의 모든 글에 많이 나오기 때문에  
핵심 내용어를 보기 어렵게 만들 수 있습니다.

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

print(list(stop_words)[:30])
```

### 설명 💡
- `stopwords.words('english')`는 영어 stopwords 목록을 불러옵니다.
- `set()`으로 바꾸면 검색이 더 편리합니다.

초보 단계에서는  
“자주 나오는 기능어 목록이구나”  
정도로 이해하면 충분합니다.

## 7. 문장부호 제거와 알파벳 단어만 남기기 🧹

tokenization 결과에는 쉼표, 마침표 같은 문장부호도 포함될 수 있습니다.  
이번에는 **알파벳으로 된 단어만** 남겨 보겠습니다.

```python
alpha_words = [word for word in tokenized_words if word.isalpha()]

print(alpha_words[:30])
print("Number of alphabetic words:", len(alpha_words))
```

### 설명 💡
- `isalpha()`는 알파벳 글자로만 이루어진 경우 `True`를 돌려줍니다.
- 따라서 이 코드는 문장부호나 숫자 등을 제외하는 데 도움이 됩니다.

## 8. stopwords 제거하기 ✂️

이제 알파벳 단어들 중에서 stopwords를 제거해 보겠습니다.

```python
content_words = [word for word in alpha_words if word not in stop_words]

print(content_words[:30])
print("Number of content words:", len(content_words))
```

### 설명 💡
이제 `the`, `and`, `of` 같은 매우 흔한 단어가 많이 줄어듭니다.  
그래서 텍스트의 **내용어(content words)** 를 조금 더 잘 볼 수 있게 됩니다.

## 9. 전처리(Pre-processing) 단계 정리 🛠️

지금까지 한 전처리 흐름은 대략 아래와 같습니다.

1. 텍스트 불러오기
2. 소문자 통일하기
3. tokenization 하기
4. 알파벳 단어만 남기기
5. stopwords 제거하기

이 흐름은 아주 기초적인 NLP 전처리 예시입니다.

```python
from nltk import FreqDist

fdist = FreqDist(content_words)
print(fdist.most_common(20))
```

### 설명 💡
- `FreqDist(content_words)`는 단어 빈도를 계산합니다.
- `most_common(20)`은 가장 자주 나오는 단어 20개를 보여 줍니다.

이제 기능어가 줄어들어서  
텍스트의 핵심 내용어가 더 잘 보일 가능성이 큽니다.

## 11. 특정 단어 빈도 확인하기 🔎

```python
print("alice:", fdist["alice"])
print("rabbit:", fdist["rabbit"])
print("queen:", fdist["queen"])
print("time:", fdist["time"])
```

이런 방식으로 특정 인물, 사물, 주제 관련 단어를 확인할 수 있습니다.

## 12. 상위 20개 단어를 표처럼 보기 📋

```python
top20 = fdist.most_common(20)

for word, freq in top20:
    print(word, ":", freq)
```

## 13. 간단한 시각화: 빈도 그래프 그리기 📊

이제 상위 단어들을 그래프로 그려 보겠습니다.

```python
import matplotlib.pyplot as plt

top10 = fdist.most_common(10)

words = [item[0] for item in top10]
freqs = [item[1] for item in top10]

plt.figure(figsize=(10, 5))
plt.bar(words, freqs)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 10 Words in Alice (after preprocessing)")
plt.xticks(rotation=45)
plt.show()
```

### 설명 💡
- `matplotlib`는 파이썬의 대표적인 시각화 라이브러리입니다.
- 막대그래프를 통해 어떤 단어가 많이 나오는지 한눈에 볼 수 있습니다.

텍스트 분석 결과를 그림으로 보는 첫 경험이라고 생각하면 됩니다 😊

## 14. NLTK의 plot 기능도 있다 📉
