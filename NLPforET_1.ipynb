## 파이썬이란?

파이썬(Python)은 사람이 비교적 읽기 쉽게 만들어진 프로그래밍 언어입니다.

예를 들어 아래 코드를 보세요.

```python
print("Hello, world!")
```

이 코드는 “Hello, world!”를 화면에 출력하라는 뜻입니다.

### 파이썬이 교육용으로 좋은 이유
- 문법이 비교적 단순합니다.
- 초보자도 빠르게 결과를 볼 수 있습니다.
- 데이터 분석, 자연어처리(NLP), 웹앱 제작까지 연결됩니다.
- ChatGPT 같은 생성형 AI와 함께 배우기 좋습니다.

## 왜 영어교사가 파이썬을 배우면 좋을까?

파이썬을 배우면 다음과 같은 일을 할 수 있습니다.

- 학생 글의 **단어 수, 문장 수, 평균 문장 길이** 계산
- 자주 나오는 단어 목록 만들기
- reading passage의 난이도 가늠하기
- cloze, vocabulary, grammar worksheet 초안 만들기
- 교사용 간단한 **웹앱** 만들기
- 생성형 AI가 만든 코드를 **읽고 수정하는 능력** 기르기

즉, 목표는 “개발자 되기”가 아니라  
**영어교육 맥락에서 쓸 수 있는 도구를 이해하고 활용하는 것**입니다.

## Google Colab이란?

Google Colab(코랩)은 **웹 브라우저에서 파이썬 코드를 실행할 수 있는 노트북 환경**입니다.

### 장점
- 설치가 거의 필요 없습니다.
- Google Drive와 연결됩니다.
- 코드와 설명을 함께 적을 수 있습니다.
- 수업 자료를 공유하기 쉽습니다.

### Colab 화면에서 학생들이 알아야 할 것
- **텍스트 셀(Text cell)**: 설명을 적는 공간
- **코드 셀(Code cell)**: 파이썬 코드를 입력하고 실행하는 공간
- **실행 버튼(▶)**: 코드 실행
- **출력(output)**: 실행 결과가 아래에 나타남

## 가장 첫 번째 코드: print()

아래 코드를 실행해 봅시다.

```python
print("안녕하세요, 파이썬!")
```

### 설명
- `print()`는 화면에 보여주는 명령입니다.
- 괄호 `()` 안에 넣은 내용을 출력합니다.
- 글자를 넣을 때는 따옴표 `" "` 또는 `' '`를 사용합니다.

## 숫자도 출력할 수 있다

```python
print(3)
print(10 + 5)
print(7 * 8)
```

### 설명
- 파이썬은 계산도 할 수 있습니다.
- `+`, `-`, `*`, `/` 같은 연산자를 사용합니다.

## 글자와 숫자는 다르다

### 설명
- `"3"`은 **문자열(string)** 입니다. 즉, 글자입니다.
- `3`은 **숫자(number)** 입니다.
- 겉보기에는 비슷하지만 파이썬에서는 다르게 처리됩니다.

## 아주 가벼운 오류 경험해 보기

아래 코드는 일부러 오류를 내는 예시입니다.

```python
print("Hello)
```

### 왜 오류가 날까요?
- 따옴표가 닫히지 않았기 때문입니다.
- 코딩에서는 **오류를 두려워하지 않고, 왜 났는지 읽는 습관**이 중요합니다.
- 오늘은 “오류가 나도 괜찮다”는 감각을 익히는 것이 중요합니다.

## 영어교육 예시 1: 텍스트 길이 보기

```python
text = "English teachers can use Python for simple text analysis."
print(text)
print(len(text))
```

### 설명
- `text`라는 이름에 문장을 저장했습니다.
- `len()`은 길이를 구해 줍니다.
- 지금은 “문자 수” 정도로 이해하면 됩니다.

## 영어교육 예시 2: 아주 간단한 앱 상상하기

```python
student_text = "I like English class because it is interesting."
print("학생 문장:", student_text)
print("문장 길이:", len(student_text))
```

이런 간단한 코드가 쌓이면 나중에는
- 글자 수 세기
- 단어 수 세기
- 자주 쓰는 단어 뽑기
- 문장 길이 분석
- reading 자료 보조

같은 기능으로 이어질 수 있습니다.

## 오늘 꼭 기억할 핵심 표현

- code cell : 코드를 실행하는 칸
- output : 실행 결과
- error : 오류
